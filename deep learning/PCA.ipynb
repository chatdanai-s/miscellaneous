```python
# Import and use iris dataset for PCA
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

iris = load_iris()
x = iris.data
y = iris.target
n, p = x.shape

print(x.shape, y.shape)
print(x)
```

```python
# Standardize x and get covariance matrix
mu = np.mean(x, axis=0)
X = x - mu
Sigma = 1/(n-1) * X.T @ X

print(mu)
print(Sigma)
```

```python
# Find eigenvectors for Sigma
eigenvalues, eigenvectors = np.linalg.eig(Sigma)

print(eigenvalues)
print(eigenvectors)
```

```python
# Get weight matrix and calculate dimenstionality-reduced parameters
Wk = eigenvectors[:,0:2]
Z = X @ Wk
print(Z)
```

```python
# Plot
Zx = Z[:,0]
Zy = Z[:,1]

plt.scatter(Zx, Zy, c=y)
plt.show()
```
